```python
import pandas as pd
```

```python
df1 = pd.DataFrame({
    'id' : [1, 2, 3, 4],
    'name' : ['Raja', ' Rani', 'Rahul', 'Rhea'],
    'age' : [25,30,35,40]
})

df2 = pd.DataFrame({
    'id' : [3, 4, 5, 6],
    'name' : ['Rahul', ' Rhea', 'Ravi', 'Reena'],
    'age' : [35,40,45,50]
})
df3 = pd.merge(df1, df2, on = 'id', how = 'inner')
df3.rename(columns = {'name_x' : 'Table_1', 'name_y' : 'Table_2'}, inplace = 'True')
df3.rename(columns = {'age_x' : 'Table_1_age', 'age_y' : 'Table_2_age'}, inplace = 'True')

display(df3)
```

```python
df4 = pd.merge(df1, df2, on = 'id', how = 'left')
print(df4)
df5 = pd.merge(df1, df2, on = 'id', how = 'right')
print(df5)
df6 = pd.merge(df1, df2, on = 'id', how = 'outer')
print(df6)
```

#Cross Join - Assuming there are x rows in table 1 and y rows in table 2, then total number of rows in resulting table would be x*y, as each row of table one would be compared to each row on the right table.

```python
df7 = pd.merge(df1, df2, how = 'cross')
print(df7)
```

#Concat - used to concatenate two tables on either axis = 0 or axis = 1


```python
df8 = pd.concat([df1, df2], axis = 0)
print(df8)
```

```python
df9 = pd.concat([df1, df2], axis = 1)
print(df9)
```

```python
df8 = pd.concat([df1, df2])
df8 = df8.drop_duplicates(subset = ['id','name'], keep  = 'first')
display(df8)
```
