```python
#List

lst = [1,2,3,'Hello','c',[2,4]]
print(lst)

#append
lst.append(96) #used to add elements at end
print(lst)

#extend
lst.extend([4,3,4]) #extend takes a new list and adds it's element to existing list, and does not addes the list structure itself.
print(lst)

#insert
lst.insert(4,3) #adds element 3 at index 4
print(lst)

#pop
lst.pop() #pops the last element of the list
print(lst)

#remove
lst.remove(4) #removes the first occurence of the chosen element from the list
print(lst)

lst.append(2)
print(lst)

#index
print(lst.index(2))

#count
print(lst.count(2))

lst2 = [34,56,4,24,52,56]
lst3 = ['abc', 'main', 'hello', 'world']
lst4 = [66,28,58,99,67,53]

#sort
lst2.sort()
print(lst2)
lst3.sort()
print(lst3)

#sum,min,max,sorted

print(sum(lst2))
print(min(lst2))
print(max(lst2))
print(sorted(lst4))

print(lst2)
print(lst4)



```

```python
#Tuples

tp = (1,2,3,78,45)
print(tp)

print(tuple(sorted(tp)))
print(tp.count(2))
print(tp.index(2))
print(len(tp))
print(sum(tp))
print(max(tp))
print(min(tp))
print(sorted(tp))
```
