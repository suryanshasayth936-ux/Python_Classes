```python
#Functions of NumPy
#zeroes and ones and full

a = np.ones((3,2,4), dtype=int)
print(a)
print(type(a[0][0][0]))

b = np.zeros((3,2,4), dtype=int)
print(b)
print(type(b[0][0][0]))

c = np.full((3,2,4),23,  dtype=int)
print(c)
print(type(c[0][0][0]))
```
