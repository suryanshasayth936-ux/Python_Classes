```python
class student:
    def __init__(self, name, age, grade):
        self.name = name
        self.age = age
        self.grade = grade

s1 = student("Alice", 15, "10th")
print(s1.name)
print(s1.grade)
```

```python

class student:
    def __init__(self, name, age, grade):
        self.name = name
        self.age = age
        self.grade = grade
    def display(self):
      print('name:',self.name)
      print('age:',self.age)
      print('grade:',self.grade)

a=student("naman",50,"NC")

a.display()
```
